## **1. Gender Distribution (Pie Chart)**
### **Insights:**
#### This chart shows the ratio of male and female students in the university.
#### It helps us understand whether the dataset is balanced or dominated by one gender.
#### If one gender is higher, it means that gender has more representation in this university dataset.
#### This is useful for comparing gender-based academic performance in further analysis.

```python
import matplotlib.pyplot as plt 
import pandas as pd
import seaborn as sns

df = pd.read_excel("Student_data.xlsx")
df["Gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.savefig("Gender Distribution")
plt.xlabel("")
plt.show()
```

## **2. Grade Distribution (Bar Chart)**
### **Insight:**
####      Most students are in average grade categories (B and C).
####      Only a small number of students achieve top grades (A+).
####      A few students are in lower grades, which shows weak performance.
####      Overall, the class performance is moderate, not extremely high or low.

```python
df["Grade"].value_counts().plot(kind="bar")
plt.savefig("Grade Distribution")
plt.title("Grade Distribution")
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.show()
```

## **3. Pass vs Fail (Pie Chart)**
### **Insight:**
#### The majority of students have passed the exams.
#### Only a small portion of students have failed.
#### This shows that overall academic results are acceptable.
#### However, failed students still need attention and support.

```python
df["Result"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    colors=["green", "red"]
)
plt.savefig("Result Ratio")
plt.title("Pass vs Fail Ratio")
plt.ylabel("")
plt.show()
```

## **4. Students By Program (Pie Chart)**
### **Insight**
#### This chart makes it easier to compare student enrollment across programs.
### Programs with longer bars have higher enrollment.
#### It clearly highlights the most and least popular programs.

```python
df["Program"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.savefig("Students by Program")
plt.title("Students by Program")
plt.ylabel("")
plt.show()
```

## **5. Attendance vs Marks (Scatter Plot)**
###  **Insight:**
#### There is a clear positive relationship between attendance and marks.
#### Students with higher attendance usually get better marks.
#### Students with low attendance mostly score lower marks.
#### This proves that attendance plays an important role in academic success.

```python
sns.scatterplot(data=df, x="Attendance", y="Obtained Marks")
plt.savefig("Attendance vs Obtained Marks")
plt.title("Attendance vs Obtained Marks")
plt.show()
```

## **6. Department-wise and Year-wise Student Performance Analysis**
#### This chart shows the average marks of students across different departments and academic years.
#### It helps compare how each department performs in each year (1st, 2nd, 3rd, 4th year).
#### Some departments consistently show higher average marks across all years, which indicates stronger academic performance and better student outcomes.
#### In contrast, a few departments show lower performance in most years, which may suggest tougher subjects, weaker student preparation, or teaching challenges.
#### Year-wise trend also shows that in many cases, higher-year students perform better than lower-year students, because they have more experience and academic maturity.
#### However, in some departments, performance does not improve significantly across years, which indicates inconsistency in learning progress.

```python
# Group data: Department + Year vs Average Marks
grouped = df.groupby(
    ["Program", "Year"]
)["Obtained Marks"].mean().unstack()

# Plot grouped bar chart
grouped.plot(kind="bar", figsize=(10, 6))
plt.savefig("Average Marks by Program and Year")
plt.title("Average Marks by Program and Year")
plt.xlabel("Program")
plt.ylabel("Average Marks")
plt.legend(title="Year")
plt.xticks(rotation=45)

plt.show()
```

## **7. Attendance Distribution**
### Insight:
#### Most students have attendance between 75% and 95%.
#### The distribution indicates a generally high level of class participation.
#### Only a small number of students have attendance below 75%.
#### High attendance suggests that students are regularly engaged in academic activities.
#### Students with better attendance are more likely to understand course material effectively.
#### Low-attendance students may face challenges in maintaining strong academic performance.
#### Overall, the dataset shows that the majority of students maintain satisfactory attendance levels, which can contribute positively to their academic success.
