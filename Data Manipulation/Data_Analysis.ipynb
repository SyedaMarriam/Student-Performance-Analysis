# **Student Performance Analysis**

## Importing Required Libraries

### In this section, we import the Python libraries required for data manipulation and analysis. Pandas is used for data handling and NumPy is used for numerical operations.

```python
# Import libraries for data manipulation
import pandas as pd
import numpy as np
```

## Loading the Dataset

### The student dataset is loaded into a Pandas DataFrame to perform data analysis and manipulation.

```python
# Load Data 
df = pd.read_excel("Student_data.xlsx")
df.head()
```

## Dataset Overview

### This section provides information about the structure of the dataset, including the number of rows, columns, and data types.

```python
# Number of rows and columns
print("Shape:", df.shape)
```

```python
# Dataset information
df.info()
```

## Descriptive Statistics
### This provides a statistical summary of numerical columns including mean, standard deviation, minimum, maximum, and quartile values.
### It helps in understanding the overall distribution and spread of the data.

```python
df.describe()
```

## Total Students
### This shows the total number of students in the dataset, defining the overall size of the analysis.

```python
print("Total Students:", len(df))
```

## Gender Distribution
### Students are distributed across male and female categories, showing the demographic balance of the dataset.

```python
print(df["Gender"].value_counts())
```

## Program-wise Student Count

### This analysis helps identify which academic program has the highest enrollment.

```python
print(df["Program"].value_counts())
```

## City-wise Student Distribution

### This analysis identifies the number of students belonging to each city. It helps understand the geographical distribution of students in the dataset.

```python
print(df["City"].value_counts())
```

## Top Performing Students

### This section identifies the highest-performing students based on obtained marks.

```python
top_students = df.nlargest(
    10,
    "Obtained Marks"
)

top_students[
    ["Name",
     "Program",
     "Obtained Marks"]
]
```

## Students Requiring Academic Attention

### This analysis identifies students with the lowest marks who may require academic support.

```python
low_students = df.nsmallest(
    10,
    "Obtained Marks"
)

low_students[
    ["Name",
     "Program",
     "Obtained Marks"]
]
```

## Pass and Fail Analysis

### Students are classified as Pass or Fail based on their obtained marks.

```python
df["Result"] = np.where(
    df["Obtained Marks"] >= 250,
    "Pass",
    "Fail"
)

df["Result"].value_counts()
```

## Program-wise Pass Rate

### This analysis evaluates the success rate of students across different programs.

```python
pd.crosstab(
    df["Program"],
    df["Result"]
)
```

## Program-wise Median GPA 
### Median GPA represents the typical performance of students in each program with minimal impact from outliers.

```python
program_gpa = df.groupby("Program")["GPA"].median()
print(program_gpa)
```

## Program-wise Mean GPA
### Mean GPA helps compare the overall academic performance across different programs.

```python
print(df.groupby("Program")["GPA"].mean())
```

## Missing Values Analysis

### Missing values can affect the quality of analysis. We identify missing values in each column before proceeding further and remove them.

```python
print(df.isnull().sum())
```

```python
df = df.dropna()   
print(df.isnull().sum())
```

## Duplicate Records Analysis

### Duplicate records can create bias in analysis. Therefore, we identify and remove duplicate entries.

```python
# Total duplicates
df.duplicated("Program").sum()
```

```python
# Remove duplicates
df = df.drop_duplicates("Program")
df.duplicated("Program").sum()
```
